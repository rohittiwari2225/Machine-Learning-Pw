Q5: What are the main differences between supervised, unsupervised, and semi-supervised learning?

Ans:-
1) Supervised Learning:-
- known input and output
- spervised learning algorithms learn from labeled data
- This means the training dataset consists of input features (X) and their corresponding correct output labels (y).
- The algorithm's goal is to learn a mapping function f:X→y so that when it sees new, unseen input data, it can predict the correct output. Think of it like learning with a teacher who provides the answers.
- Supervised Learning: The primary goal is prediction
- The model learns to predict the output label for new, unseen input data based on the patterns learned from the labeled training data. This can be either predicting a continuous value (regression) or a discrete category (classification).
>  use of Supervised Learning:
Image classification (identifying objects in images)
Spam detection (classifying emails as spam or not)
Regression (predicting house prices, stock prices)
Speech recognition (converting audio to text)
Sentiment analysis (determining the emotion in text)


2) Unsupervised Learning
- unsupervised learning algorithms learn from unlabeled data.
- The training data only contains input features (X) without any corresponding output labels.
-  The algorithm's goal is to find hidden patterns, structures, or relationships within the data itself. It's like learning without a teacher, trying to make sense of the data on its own.
- Unsupervised Learning: The main goals are discovery and understanding. This involves tasks like:
Clustering: Finding natural groupings within the data.
Dimensionality Reduction: Reducing the number of features while preserving important information.
Association Rule Mining: Discovering relationships between different variables.
Anomaly Detection: Identifying unusual data points.
Generative Modeling: Learning the data distribution to create new, similar data.

> use of unsupervised learning
- Customer segmentation (grouping customers based on behavior)
- Anomaly detection (identifying fraudulent transactions)
- Document clustering (grouping similar documents)
- Dimensionality reduction (e.g., PCA for data visualization)
- Market basket analysis (finding associations between purchased items)


3) Semi-supervised Learning 
- This approach is a hybrid. It uses a combination of a small amount of labeled data and a large amount of unlabeled data for training
- The idea is that the large amount of unlabeled data can help to understand the underlying structure of the data, while the small amount of labeled data provides guidance for the learning process.
- Semi-Supervised Learning: The goal is to improve learning performance by leveraging the large amount of unlabeled data in addition to the small amount of labeled data. This can often lead to better accuracy and generalization than using only the small labeled dataset. It aims to get the benefits of both supervised and unsupervised learning.

> use of semi supervised learning
- Text document classification (classifying a large corpus of documents with only a few labeled examples)
- Image classification (improving image recognition with a small labeled dataset and a larger unlabeled one)
- Speech analysis (training speech recognition models with limited labeled audio)
Medical diagnosis (classifying medical images with a limited number of labeled scans)




Q6: What is train, test and validation split? Explain the importance of each term.
 
Ans:-
1. Training Set:

- What it is: This is the largest portion of your dataset (often 70-80%). It's the data that your machine learning model learns from. The algorithm looks at the input features and their corresponding output labels (in supervised learning) or the input features themselves (in unsupervised learning) and adjusts its internal parameters (weights, biases, etc.) to find patterns and relationships.
- Importance: The training set is fundamental because it's where the model develops its understanding of the underlying data distribution and builds its predictive capabilities. Without a sufficiently large and representative training set, the model might not learn the relevant patterns or might overfit to the specific nuances of the training data.
2. Validation Set (or Development Set):

- What it is: This is a smaller portion of your dataset (often 10-15%) that is held back from the training process. It's used to evaluate the model's performance during training and to tune hyperparameters. Hyperparameters are settings of the learning algorithm that are not learned from the data itself (e.g., the learning rate, the number of layers in a neural network, the complexity of a decision tree).
- Importance: The validation set plays a critical role in:
Hyperparameter Tuning: By evaluating the model's performance on the validation set with different hyperparameter settings, you can choose the combination that leads to the best generalization. You're essentially preventing yourself from overfitting to the training data while still optimizing the model's architecture and learning process.
- Early Stopping: You can monitor the model's performance on the validation set during training. If the performance starts to degrade (indicating overfitting to the training data), you can stop the training process early, before the model learns the noise in the training data.
- Model Selection: If you're experimenting with different models, you can use the validation set to compare their performance and choose the one that generalizes best.
3. Test Set (or Hold-Out Set):

- What it is: This is the final, and usually the smallest, portion of your dataset (often 10-15%) that is completely held back until the very end of the model development process. The model never sees the test set during training or hyperparameter tuning.
- Importance: The test set provides a final, unbiased evaluation of the trained model's performance on completely new, unseen data. It simulates how well the model will perform in the real world. The results on the test set are what you typically report as the model's generalization ability. It helps to avoid:
- Overestimation of Performance: If you evaluate your model only on the training data, you'll likely get overly optimistic results because the model has already "seen" and learned from that data. The test set provides a more realistic estimate of performance.
- Information Leakage: Using the test set for hyperparameter tuning or model selection would introduce information leakage from the test set into the model building process, leading to an overly optimistic and unreliable evaluation.
